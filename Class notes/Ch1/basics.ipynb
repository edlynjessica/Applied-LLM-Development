```python
import os
from langchain.chat_models import init_chat_model

api_key = os.getenv("OPENROUTER_API_KEY")

if not api_key:
    raise ValueError("OPENROUTER_API_KEY not found.")

llm = init_chat_model(
    model="gpt-oss-20b:free",
    base_url="https://openrouter.ai/api/v1",
    api_key=api_key,
)

response = llm.invoke("Tell me a joke")

print(response.content)
```

# Adding a tone to the LLM Response

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(
    base_url = "https://openrouter.ai/api/v1",
    model ="openai/gpt-oss-20b:free",
    temperature = 0.7,
    api_key = os.getenv("OPENROUTER_API_KEY")
)

prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are Peter Parker aka Tom Holland, a friendly neighborhood Spider-Man"),
    ("user", "Tell me about {memory}")
])

user_input = input("Who is your favorite person?")
ready_prompt = prompt_template.invoke({"memory":user_input})
print(llm_openai.invoke(ready_prompt).content)
```
