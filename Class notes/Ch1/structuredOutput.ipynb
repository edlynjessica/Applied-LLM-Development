# 📧 Structured Output with Pydantic

Instead of receiving a plain text response from an LLM, **Structured Output** ensures the response follows a predefined format. This makes the output easier to validate, access, and integrate into applications.

---

## 🎯 What is Structured Output?

Structured Output allows the LLM to return data in a specific schema rather than free-form text.

For example, instead of:

> Subject: Medical Leave Request  
> Dear HR,...

the model returns an object with clearly defined fields:

- 📌 **Subject**
- 📝 **Email Content**

This eliminates the need for manually parsing the response.

---

## 🏗️ Why Use Pydantic?

**Pydantic** is a Python library used for **data validation** and **schema definition**.

It helps by:

- ✅ Defining the expected structure of the response.
- ✅ Validating that the response matches the schema.
- ✅ Providing easy access to individual fields.
- ✅ Reducing errors caused by inconsistent LLM outputs.

---

## 📦 Creating a Schema

A **Pydantic model** acts as a blueprint for the expected response.

For an email, the schema might contain:

- 📌 `subject`
- 📝 `email_content`

The LLM is instructed to generate responses that match this structure.

---

## 🤖 Structured Output in LangChain

LangChain allows you to bind a Pydantic model to an LLM using **Structured Output**.

Once attached, the model no longer returns plain text—it returns an object that follows the predefined schema.

Instead of accessing text manually, you can directly retrieve individual fields like:

- 📌 Subject
- 📝 Email Body

This makes the response predictable and developer-friendly.

---

## 🏷️ What is `Field()`?

`Field()` is used to add **metadata** to each attribute in the schema.

It can provide:

- 📖 Descriptions
- 📌 Default values
- 🔒 Validation rules
- 📏 Constraints

Example use cases include specifying that a field should contain an email subject or limiting the maximum length of a string.

---

## 💡 Why Add Descriptions?

Field descriptions help the LLM better understand the purpose of each attribute.

Instead of simply knowing that a field is called `subject`, the model also understands that it represents **the subject line of an email**.

This often leads to more accurate and consistent structured responses.

---

## ✅ Benefits of Structured Output

- 📄 Produces predictable responses.
- 🛡️ Validates the generated data automatically.
- 🎯 Eliminates manual text parsing.
- ⚡ Makes responses easier to use in applications.
- 🔄 Integrates seamlessly with APIs, databases, and automation workflows.

---

## 📌 Key Takeaways

- 🤖 **Structured Output** returns data in a predefined format instead of plain text.
- 🏗️ **Pydantic** defines and validates the expected response schema.
- 📦 A **BaseModel** acts as the blueprint for the LLM's output.
- 🏷️ **Field()** adds descriptions, defaults, and validation rules to model attributes.
- 🚀 Structured Output makes LLM responses more reliable, consistent, and easier to process programmatically.

```python
import os
from dotenv import load_dotenv

load_dotenv()

if os.getenv("OPENROUTER_API_KEY") is not None:
    print("OpenRouter API Key Found")
else:
    print("OpenRouter API Key Not Found")
from langchain_openai import ChatOpenAI

llm_openai = ChatOpenAI(base_url = "https://openrouter.ai/api/v1", model ="openai/gpt-oss-20b:free", temperature = 0.7, api_key = os.getenv("OPENROUTER_API_KEY"))

## **Using Pydantic for Structured Output**
from pydantic import BaseModel

class email(BaseModel):
    subject: str
    email_content: str


email(**{"subject": "sample subject", "email_content": "sample mail content"})

llm_openai_new = llm_openai.with_structured_output(email)
response = llm_openai_new.invoke("Generate subject and email for applying medical leave to my HR")
response
response.subject
print(response.email_content)
from pydantic import BaseModel, Field

class email(BaseModel):
    subject: str = Field(description="The subject of the mail.")
    email_content: str = Field(description="The body content of the email.")
```
